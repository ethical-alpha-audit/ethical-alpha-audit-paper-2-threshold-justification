# Notebook 04: Release Validation

**Paper 3:** Beyond 'Magic Numbers': A Threshold Justification Stack for Clinical AI Governance

This notebook validates the repository structure, data integrity, and output consistency. It should be the last notebook executed and should pass without errors.

```python
import json
from pathlib import Path

REPO_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()

checks_passed = 0
checks_failed = 0

def check(label, condition):
    global checks_passed, checks_failed
    if condition:
        checks_passed += 1
        print(f'  PASS: {label}')
    else:
        checks_failed += 1
        print(f'  FAIL: {label}')
```

## Directory Structure

```python
expected_dirs = [
    'manuscript', 'data/tables', 'data/schemas', 'notebooks',
    'outputs/tables', 'outputs/schemas', 'config', 'scripts',
    'docs', 'tests', 'logs'
]
print('Directory structure checks:')
for d in expected_dirs:
    check(d, (REPO_ROOT / d).is_dir())
```

## Manuscript Files

```python
print('Manuscript file checks:')
check('Paper3_Manuscript.docx', (REPO_ROOT / 'manuscript' / 'Paper3_Manuscript.docx').exists())
check('Paper3_Supplementary_Materials.docx', (REPO_ROOT / 'manuscript' / 'Paper3_Supplementary_Materials.docx').exists())
```

## Data File Integrity

```python
data_table_files = [
    'table1_failure_mechanisms.json', 'table2_tjs_specification.json',
    'table3_framework_comparison.json', 'glossary.json',
    'nhs_governance_mapping.json', 'gaming_resistance_pathways.json'
]
print('Data table file checks (existence + valid JSON):')
for fname in data_table_files:
    fpath = REPO_ROOT / 'data' / 'tables' / fname
    exists = fpath.exists()
    valid_json = False
    if exists:
        try:
            json.loads(fpath.read_text(encoding='utf-8'))
            valid_json = True
        except (json.JSONDecodeError, Exception):
            pass
    check(f'{fname} exists', exists)
    check(f'{fname} valid JSON', valid_json)

schema_files = ['tjs_record_primary_safety.json', 'tjs_record_secondary_operational.json']
print('\nSchema file checks (existence + valid JSON):')
for fname in schema_files:
    fpath = REPO_ROOT / 'data' / 'schemas' / fname
    exists = fpath.exists()
    valid_json = False
    if exists:
        try:
            data = json.loads(fpath.read_text(encoding='utf-8'))
            valid_json = True
            check(f'{fname} has _epistemic_status', '_epistemic_status' in data)
        except (json.JSONDecodeError, Exception):
            pass
    check(f'{fname} exists', exists)
    check(f'{fname} valid JSON', valid_json)
```

## Output File Checks

```python
output_table_files = [
    'table1_failure_mechanisms.csv', 'table2_tjs_specification.csv',
    'table3_framework_comparison.csv', 'glossary.csv',
    'nhs_governance_mapping.csv', 'gaming_resistance_pathways.csv'
]
print('Output table checks:')
for fname in output_table_files:
    fpath = REPO_ROOT / 'outputs' / 'tables' / fname
    exists = fpath.exists()
    nonempty = exists and fpath.stat().st_size > 0
    check(f'{fname} exists and non-empty', nonempty)

output_schema_files = [
    'tjs_record_primary_safety_rendered.json',
    'tjs_record_secondary_operational_rendered.json'
]
print('\nOutput schema checks:')
for fname in output_schema_files:
    fpath = REPO_ROOT / 'outputs' / 'schemas' / fname
    exists = fpath.exists()
    nonempty = exists and fpath.stat().st_size > 0
    check(f'{fname} exists and non-empty', nonempty)
```

## Config File Checks

```python
config_files = ['harness_settings.json', 'notebook_plan.json', 'expected_outputs.json', 'trace_map.json']
print('Config file checks:')
for fname in config_files:
    fpath = REPO_ROOT / 'config' / fname
    exists = fpath.exists()
    valid_json = False
    if exists:
        try:
            json.loads(fpath.read_text(encoding='utf-8'))
            valid_json = True
        except (json.JSONDecodeError, Exception):
            pass
    check(f'{fname} exists and valid', exists and valid_json)
```

## Notebook Count

```python
nb_count = len(list((REPO_ROOT / 'notebooks').glob('*.ipynb')))
print(f'Notebooks found: {nb_count}')
check('Exactly 4 notebooks', nb_count == 4)
```

## Summary

```python
total = checks_passed + checks_failed
print(f'\n{"=" * 50}')
print(f'Total checks: {total}')
print(f'Passed: {checks_passed}')
print(f'Failed: {checks_failed}')
print(f'{"=" * 50}')
if checks_failed == 0:
    print('RELEASE VALIDATION: PASS')
    print('Repository is structurally complete and ready for release.')
else:
    print('RELEASE VALIDATION: FAIL')
    print(f'{checks_failed} check(s) failed. Review above for details.')
assert checks_failed == 0, f'{checks_failed} validation check(s) failed'
```
