## The Threshold Justification Stack

Six documentation elements are specified for TJS-compliant governance records. The tiered architecture requires:
- **Primary Safety Thresholds** (failure directly causes patient harm): all 6 layers
- **Secondary Operational Thresholds** (failure affects operations, not direct patient harm): 3 core layers

```python
with open(REPO_ROOT / 'data' / 'tables' / 'table2_tjs_specification.json', 'r', encoding='utf-8') as f:
    t2_data = json.load(f)

df_t2 = pd.DataFrame(t2_data['data'])
df_t2.columns = ['Layer #', 'Layer Name', 'Description', 'Mechanism Addressed', 'Tier Requirement', 'Regulatory Counterpart']
df_t2
```

```python
out_path = REPO_ROOT / 'outputs' / 'tables' / 'table2_tjs_specification.csv'
df_t2.to_csv(out_path, index=False, encoding='utf-8')
print(f'Exported: {out_path.relative_to(REPO_ROOT)}')
```

**Caveat (from manuscript):** All regulatory counterpart mappings are the author's interpretive inferences reflecting the governance intent of the cited provisions. They should not be read as statements of regulatory intent or legal requirement.

## Tier Classification Logic

The TJS defines three classification categories for governance thresholds:

```python
tier_rules = pd.DataFrame([
    {'Tier': 'Primary Safety', 'Definition': 'Failure can directly cause patient harm', 'Layers Required': 'All 6', 'Default Rule': '—'},
    {'Tier': 'Secondary Operational', 'Definition': 'Failure does not directly cause irreversible patient harm', 'Layers Required': '3 core (L1 recommended, L2 required, L6 required; L3-L5 optional)', 'Default Rule': '—'},
    {'Tier': 'Uncertain', 'Definition': 'Classification cannot be determined', 'Layers Required': 'All 6 (defaults to Primary Safety)', 'Default Rule': 'Defaults to Primary Safety requirements'},
])
tier_rules
```

The "uncertain" classification option provides a conservative pathway for ambiguous cases. The reliability of tier classification across different assessors is an empirical question that this conceptual paper does not resolve — see the proposed inter-rater reliability exercise in the Discussion.
