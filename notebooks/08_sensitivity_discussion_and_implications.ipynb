## 08.1 From regime to falsification: linking sensitivity to the manuscript's pilot design

The κ-sensitivity simulation in notebooks 05–07 establishes that under stated modelling assumptions, **κ < 0.40 is the falsification regime** — the regime in which the Negative Harm Test cannot defensibly support the framework's tier-stratified documentation depth. This regime boundary maps directly onto the manuscript's Discussion of falsification conditions.

The connection: the manuscript's Discussion §"Theory of change and falsification conditions" identifies as one falsification condition that **the proposed pilot inter-rater reliability exercise produces κ < 0.40 on Primary/Secondary tier classification** (P3-C45). This is a condition for the framework's failure: if a real pilot were to measure κ < 0.40 in an institutional setting, the proportionality argument that motivates the tiered TJS would no longer be defensible at that institution.

Why this falsification condition is interpretable: **the first falsification condition is interpretable as a regime with quantified analytical consequences** (see Results §κ regimes) (P3-C46). The simulation has shown that the κ<0.40 regime corresponds to specific quantified governance outcomes (unsafe-Secondary rates of 1.5%–2.8% at π=0.10, with monotonic and structural variation across the κ × π grid). So a real-pilot result in the falsification regime is not just a number; it is a specific, quantified prediction about what the rule would produce in operation.

The simulation provides the **interpretive bridge** between the falsification κ<0.40 condition (a measurable outcome of a hypothetical empirical exercise) and its analytical consequences (the rule produces non-trivial misclassification rates in operation). The cross-reference: the visible reference below shows the falsification-regime data from notebook 06's verified regime-ordering result.

```python
# Visible reference: the falsification regime's quantified analytical consequences
# (display-only; no new computation. Loads the same sensitivity_main_pi_010.csv as notebook 06.
# The regime-ordering assertion that establishes the κ<0.40 boundary is verified in notebook 06.)
import csv

results_path = REPO_ROOT / SENSITIVITY_TABLES_DIR / "sensitivity_main_pi_010.csv"
with open(results_path) as f:
    rows = list(csv.DictReader(f))

print("Falsification-regime data (κ < 0.40, π = 0.10):")
print(f"{'target κ':>10} {'unsafe-Secondary rate':>22} {'MCSE':>10}")
for r in rows:
    k = float(r['target_kappa'])
    if k < 0.40:
        rate = float(r['unsafe_secondary_rate_mean'])
        mcse = float(r['unsafe_secondary_rate_mcse'])
        print(f"{k:>10.2f} {rate:>22.5f} {mcse:>10.5f}")
print()
print("Reference: notebook 06 assert 2 verifies regime ordering")
print("(falsification mean > caution mean > operational mean for unsafe-Secondary rate at π=0.10).")
print("This notebook references that verified result; it does not re-assert it.")
```

## 08.4 Anti-gaming acknowledgements

The manuscript acknowledges a key risk to the framework's operational integrity: **documentation artefacts can be produced as compliance theatre — records completed for each threshold but without substantive engagement with the harm-control link or the sensitivity analysis** (P3-C56). This is the framework's anti-gaming limitation: a TJS record can be filled in mechanically (committee signs off, layers populated with boilerplate) without the underlying threshold-justification reasoning actually being done.

This risk is structural to any documentation framework. The TJS does not solve it — it acknowledges it and provides specific points where gaming is possible:

- **Tier misclassification**: a Primary Safety threshold deliberately classified Secondary to reduce documentation burden (the gaming-resistance pathways in `outputs/tables/gaming_resistance_pathways.csv` from the framework-comparison work, notebook 02, address this explicitly).
- **Sensitivity analysis (Layer 4) shortcut**: the layer's documentation can be populated with formulaic content rather than substantive sensitivity exploration.
- **Harm-control link (Layer 1) shortcut**: the link can be asserted without justification.

The simulation cannot detect or prevent compliance theatre — that is a behavioural and institutional concern, not a computational one. The κ regime work in notebooks 05–07 does, however, provide the empirical basis for one form of gaming detection: if a TJS audit reveals that an institution's tier classifications cluster heavily in one direction (e.g. nearly all thresholds classified Secondary to minimise documentation), the κ regime work provides quantified evidence about what the operational consequences would be under different reliability assumptions.

This claim has no computational handle in this notebook. It earns **Traced** status.
