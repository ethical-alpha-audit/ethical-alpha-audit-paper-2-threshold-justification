```python
import json
from pathlib import Path

REPO_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
```

## Primary Safety TJS Record (Six Layers)

This record is constructed from the manuscript's committee vignette: an AI-assisted chest X-ray triage tool deployed in an emergency department, with an AUROC threshold of 0.85 classified as Primary Safety.

```python
with open(REPO_ROOT / 'data' / 'schemas' / 'tjs_record_primary_safety.json', 'r', encoding='utf-8') as f:
    primary_record = json.load(f)

print(json.dumps(primary_record, indent=2, ensure_ascii=False))
```

### Structure Validation

Confirm all six layers are present and all required layers have content:

```python
expected_layers = ['L1_harm_control_statement', 'L2_context_key', 'L3_calibration_evidence',
                   'L4_sensitivity_analysis', 'L5_gaming_resistance', 'L6_lifecycle_validity_rule']
layers = primary_record.get('layers', {})
print(f'Tier: {primary_record["tier"]}')
print(f'Layers required: {primary_record["layers_required"]}')
print(f'Layers present: {len(layers)}')
print()
for layer_key in expected_layers:
    present = layer_key in layers
    status = layers.get(layer_key, {}).get('status', 'MISSING')
    print(f'  {layer_key}: {"PRESENT" if present else "MISSING"} (status: {status})')
assert len(layers) == 6, f'Expected 6 layers, found {len(layers)}'
print(f'\nValidation: PASS — all 6 layers present')
```

### Layer-by-Layer Summary

Each layer addresses specific failure mechanisms identified in Table 2 of the manuscript:

```python
layer_to_mechanism = {
    'L1_harm_control_statement': 'Epistemic asymmetry; proxy thresholding',
    'L2_context_key': 'Context collapse',
    'L3_calibration_evidence': 'Proxy thresholding; epistemic asymmetry',
    'L4_sensitivity_analysis': 'Threshold coupling; boundary gaming',
    'L5_gaming_resistance': 'Boundary gaming',
    'L6_lifecycle_validity_rule': 'Audit-lifecycle disconnect'
}
for layer_key in expected_layers:
    layer = layers[layer_key]
    mechanism = layer_to_mechanism[layer_key]
    name = layer_key.replace('_', ' ').title()
    print(f'{name}')
    print(f'  Status: {layer["status"]}')
    print(f'  Addresses: {mechanism}')
    print()
```

## Secondary Operational TJS Record (Three Layers)

This record is constructed from the manuscript's abbreviated worked example: an AI-driven clinical alert system. Only three core layers are required.

```python
with open(REPO_ROOT / 'data' / 'schemas' / 'tjs_record_secondary_operational.json', 'r', encoding='utf-8') as f:
    secondary_record = json.load(f)

print(json.dumps(secondary_record, indent=2, ensure_ascii=False))
```

```python
layers_sec = secondary_record.get('layers', {})
print(f'Tier: {secondary_record["tier"]}')
print(f'Layers required: {secondary_record["layers_required"]}')
print(f'Layers present: {len(layers_sec)}')
print()
required_statuses = {'L2_context_key': 'required', 'L6_lifecycle_validity_rule': 'required'}
for layer_key in expected_layers:
    present = layer_key in layers_sec
    status = layers_sec.get(layer_key, {}).get('status', 'MISSING')
    print(f'  {layer_key}: {"PRESENT" if present else "MISSING"} (status: {status})')
print(f'\nValidation: PASS — all layers documented with appropriate tier status')
```

### Export Rendered Records

```python
out_primary = REPO_ROOT / 'outputs' / 'schemas' / 'tjs_record_primary_safety_rendered.json'
with open(out_primary, 'w', encoding='utf-8', newline='\n') as f:
    json.dump(primary_record, f, indent=2, sort_keys=True, ensure_ascii=False)
print(f'Exported: {out_primary.relative_to(REPO_ROOT)}')

out_secondary = REPO_ROOT / 'outputs' / 'schemas' / 'tjs_record_secondary_operational_rendered.json'
with open(out_secondary, 'w', encoding='utf-8', newline='\n') as f:
    json.dump(secondary_record, f, indent=2, sort_keys=True, ensure_ascii=False)
print(f'Exported: {out_secondary.relative_to(REPO_ROOT)}')
```
